```python
from rdkit import Chem
from rdkit.Chem import rdShapeHelpers

ligands = {
    "10176710": Chem.MolFromMolFile("10176710_rep1.sdf", removeHs=False),
    "m5": Chem.MolFromMolFile("m5_rep1.sdf", removeHs=False),
    "camp": Chem.MolFromMolFile("cAMP_rep1.sdf", removeHs=False),
    "rolR": Chem.MolFromMolFile("rolR_rep1.sdf", removeHs=False),
}

for name, mol in ligands.items():
    if mol is None:
        raise ValueError(f"Não foi possível carregar o ligante: {name}")

names = list(ligands.keys())

for i, name1 in enumerate(names):
    for name2 in names[i + 1:]:
        dist = rdShapeHelpers.ShapeTanimotoDist(ligands[name1], ligands[name2])
        similarity = 1 - dist
        print(f"{name1} x {name2}: dist={dist:.4f}, simil={similarity:.4f}")
```
