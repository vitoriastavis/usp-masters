```python
!pip install rdkit
from rdkit import Chem
from rdkit.Chem import AllChem

# Estrutura química original
template = Chem.SDMolSupplier("10394.sdf")[0]
template = Chem.RemoveHs(template)
    
# Pose do docking
pose = Chem.MolFromPDBFile(
    "10394_docked.pdb",
    removeHs=True,
    sanitize=False
)

# Recuperar ordens de ligação
ligand = AllChem.AssignBondOrdersFromTemplate(template, pose)

# Adicionar hidrogênios
ligand = Chem.AddHs(ligand, addCoords=True)

# Salvar estrutura reconstruída
Chem.MolToMolFile(ligand, "10394_prepared.sdf")
```
