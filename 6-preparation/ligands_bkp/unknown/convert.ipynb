```python
from rdkit import Chem
from rdkit.Chem import AllChem

smiles = ["C(CCC1=CC(O)=C(O)C=C1)(=O)C(O)C(O)CO",
          "O[C@H](CCC(=O)O)CC1=CC(=C(C=C1)O)O",
          "O[C@H](CCC(=O)O)CC1=CC(=CC=C1)O",
          "O[C@@H](CCC(=O)O)CC1=CC(=CC=C1)O"
          ]

names = ["no7", "no12", "no13", "no17"]

for i in range(len(smiles)):
    ligand = smiles[i]
    print(ligand)
    name = f"{names[i]}.pdb"

    mol = Chem.MolFromSmiles(ligand)
    mol = Chem.AddHs(mol)

    AllChem.EmbedMolecule(mol)
    AllChem.MMFFOptimizeMolecule(mol)

    Chem.MolToPDBFile(mol, name)
```
